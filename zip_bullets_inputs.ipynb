```python
import zipfile
import os
```

```python
# Put bullets_....txt files into archive (only if archive doesn't already exist)
if not os.path.isfile("input/bullets_base.zip"):
    with zipfile.ZipFile("input/bullets_base.zip", "w", compression=zipfile.ZIP_DEFLATED) as zip_file:
        for f in [f for f in os.listdir('input')
                if f.startswith('bullets') and f.endswith('.txt')]:
            zip_file.write(os.path.join('input', f))
```

```python
# Pull bullets_....txt files from archive
with zipfile.ZipFile("input/bullets_base.zip", "r") as zip_file:
    zip_file.extractall("input")
```
